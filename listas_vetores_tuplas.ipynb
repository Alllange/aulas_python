```python
vendas.append(2500)
print("Vendas após adicionar um novo valor:", vendas)
```

```python
#adicionar em posição específica: insert()

vendas.insert(1, 5000)

print(vendas)
```

```python
# adicionar vários elemntos extend()

vendas.extend([10,20, 30])
vendas
```

```python
novos_elementos = [40, 50, 60]

vendas.extend(novos_elementos)

vendas
```

```python
# REMOVER UM ELEMENTO PELO O INDEX POP()

vendas.pop(0)# Remove o primeiro elemento

vendas
```

```python
# ordenar uma lista, função sort() - ordena a lista em ordem crescente
vendas.sort() # Ordena a lista em ordem crescente

vendas
```

```python
vendas.sort(reverse=True) # Ordena a lista em ordem decrescente

print(vendas)
```

```python
# fatiamento de listas


vendas[1:3] # Retorna os elementos do índice 1 até o índice 2 (3 não é incluído)

print(vendas)
```
